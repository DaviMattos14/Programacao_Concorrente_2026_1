## Atividade 3 — Paralelização Soma de Vetor

**Objetivo:** Paralelizar, com uso apenas das diretivas `parallel`, `master` e `single` o programa a seguir.

### Programa: `omp_soma_vetor.c`


```python
%%writefile omp_soma_vetor.c

#include <stdio.h>
#include <stdlib.h>
#include <omp.h>

int main() {
    int N = 1000000;
    double *A = malloc(sizeof(double) * N);

    for (int i = 0; i < N; i++) {
        A[i] = 1.0;
    }

    double soma = 0.0;

    for (int i = 0; i < N; i++) {
        soma += A[i];
    }

    printf("Soma = %.0f\n", soma);
    free(A);
    return 0;
}
```

```python
!gcc -o teste omp_soma_vetor.c -fopenmp
!./teste
```

## Atividade 4 — Paralelização Método do Trapézio

**Objetivo:** Paralelizar, com uso apenas das diretivas `parallel`, `master` e `single` o programa a seguir.

### Programa: `omp_trapezio.c`


```python
%%writefile omp_trapezio.c

#include <stdio.h>
#include <math.h>

double f(double x) {
    return exp(x);
}

int main() {
    double a = 0.0, b = 1.0;
    int n = 10000000;
    double h = (b - a) / n;
    double soma = 0.5 * (f(a) + f(b));

    for (int i = 1; i < n; i++) {
        soma += f(a + i * h);
    }

    printf("Integral = %.6f\n", soma * h);
    return 0;
}
```

```python
!gcc -o teste omp_trapezio.c -fopenmp -lm
print("Executando 4 vezes para verificar consistência do resultado:")
for i in range(4):
    print(f"--- Execução {i+1} ---")
    !./teste
    print()
```
